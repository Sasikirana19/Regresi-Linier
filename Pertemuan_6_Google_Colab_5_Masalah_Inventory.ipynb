```r
H<-0.1
C0<-15
data.d<-c( 12,8,14,13,21,14,7,14,8,9,12,12,12,11,13,8,13,11,15,12,
           10,13,9,12,7,13,10,18,9,11,10,15,12,13,7,11,17,9,15,7,
           13,13,12,14,17,10,10,14,11,17,12,13,8,8,6,15,9,9,13,16)
```

```r
md<-mean(data.d)
print("md")
print(md)
```

```r
isi<-(2*C0*md)/H
EOQ<-sqrt(isi)
print("EOQ")
print(EOQ)
```

```r
T<-EOQ/md
print("T")
print(T)
```

```r
# simulasi 1

#Q<-ceiling(EOQ)
Q<-floor(EOQ)
T1<-ceiling(T)
#T1<-floor(T)
#T1<-ceiling(T)+2
LoI<-c(Q)
akhir<-Q
siklus<-1

for (i in c(1:length(data.d))){
  =if (i=siklus*T1){
    siklus<-siklus+1
    if (akhir<0){LoI<-c(LoI,Q)
      akhir<-Q
    }else{
      akhir<-akhir+Q
      LoI<-c(LoI,akhir)
    }
  }else{
    cek_akhir<-akhir-data.d[i]
    LoI<-c(LoI,cek_akhir)
    akhir<-cek_akhir
  }
}
```

```r
barplot(LoI)
```

```r
# biaya penyimpanan
holding<-0
for (i in c(1:T1)){
  if (LoI[i]>0){holding<-holding+LoI[i]*H}
}

print(holding)
```

```r
biaya<-biaya+holding
print(biaya)
```
