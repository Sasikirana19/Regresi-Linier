```python

import time

ALFABET = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"


# ==========================================
# 1. FUNGSI CAESAR
# ==========================================

def enkripsi_caesar(teks, geser):
    hasil = ""

    for huruf in teks:
        posisi = ALFABET.index(huruf)
        posisi_baru = (posisi + geser) % 26
        hasil += ALFABET[posisi_baru]

    return hasil


def dekripsi_caesar(teks, geser):
    return enkripsi_caesar(teks, -geser)


# ==========================================
# 2. FUNGSI VIGENERE
# ==========================================

def enkripsi_vigenere(teks, kunci):
    hasil = ""

    for i, huruf in enumerate(teks):
        posisi_teks = ALFABET.index(huruf)
        posisi_kunci = ALFABET.index(kunci[i % len(kunci)])
        posisi_baru = (posisi_teks + posisi_kunci) % 26
        hasil += ALFABET[posisi_baru]

    return hasil


def dekripsi_vigenere(teks, kunci):
    hasil = ""

    for i, huruf in enumerate(teks):
        posisi_teks = ALFABET.index(huruf)
        posisi_kunci = ALFABET.index(kunci[i % len(kunci)])
        posisi_baru = (posisi_teks - posisi_kunci) % 26
        hasil += ALFABET[posisi_baru]

    return hasil


# ==========================================
# 3. FUNGSI TRANSPOSISI KOLOM
# ==========================================

def enkripsi_transposisi(teks, kunci):
    jumlah_kolom = len(kunci)

    urutan_kolom = sorted(
        range(jumlah_kolom),
        key=lambda i: kunci[i]
    )

    hasil = ""

    for kolom in urutan_kolom:
        for i in range(kolom, len(teks), jumlah_kolom):
            hasil += teks[i]

    return hasil


def dekripsi_transposisi(teks, kunci):
    jumlah_kolom = len(kunci)
    panjang = len(teks)

    jumlah_baris = panjang // jumlah_kolom
    sisa = panjang % jumlah_kolom

    urutan_kolom = sorted(
        range(jumlah_kolom),
        key=lambda i: kunci[i]
    )

    kolom = [""] * jumlah_kolom
    posisi = 0

    for nomor_kolom in urutan_kolom:
        panjang_kolom = jumlah_baris

        if nomor_kolom < sisa:
            panjang_kolom += 1

        kolom[nomor_kolom] = teks[
            posisi:posisi + panjang_kolom
        ]

        posisi += panjang_kolom

    hasil = ""

    for baris in range(jumlah_baris + 1):
        for nomor_kolom in range(jumlah_kolom):
            if baris < len(kolom[nomor_kolom]):
                hasil += kolom[nomor_kolom][baris]

    return hasil


# ==========================================
# 4. FUNGSI AFFINE
# ==========================================

def enkripsi_affine(teks, a, b):
    hasil = ""

    for huruf in teks:
        posisi = ALFABET.index(huruf)
        posisi_baru = (a * posisi + b) % 26
        hasil += ALFABET[posisi_baru]

    return hasil


def dekripsi_affine(teks, a, b):
    invers_a = 0

    for i in range(26):
        if (a * i) % 26 == 1:
            invers_a = i

    if invers_a == 0:
        return "Kunci Affine tidak valid"

    hasil = ""

    for huruf in teks:
        posisi = ALFABET.index(huruf)
        posisi_baru = (invers_a * (posisi - b)) % 26
        hasil += ALFABET[posisi_baru]

    return hasil


# ==========================================
# 5. KUNCI RAHASIA KELOMPOK
# ==========================================

kunci_caesar = 17
kunci_vigenere = "INIKUNCINYA"
kunci_transposisi = "ICNUK"
kunci_a = 5
kunci_b = 8


# ==========================================
# 6. PLAINTEXT
# ==========================================

plaintext = "RAHASIAKELOMPOKANDAADALAH"


# ==========================================
# 7. PROSES ENKRIPSI
# ==========================================

print("===================================")
print("          PROSES ENKRIPSI")
print("===================================")

waktu_mulai = time.time()

cipher1 = enkripsi_caesar(
    plaintext, kunci_caesar
)

cipher2 = enkripsi_vigenere(
    cipher1, kunci_vigenere
)

cipher3 = enkripsi_transposisi(
    cipher2, kunci_transposisi
)

cipher_final = enkripsi_affine(
    cipher3, kunci_a, kunci_b
)

waktu_enkripsi = time.time() - waktu_mulai

print("Plaintext          :", plaintext)
print("Hasil Caesar       :", cipher1)
print("Hasil Vigenere     :", cipher2)
print("Hasil Transposisi  :", cipher3)
print("Ciphertext akhir   :", cipher_final)
print("Waktu enkripsi     :", waktu_enkripsi, "detik")


# ==========================================
# 8. PROSES DEKRIPSI
# ==========================================

print("\n===================================")
print("          PROSES DEKRIPSI")
print("===================================")

waktu_mulai = time.time()

hasil1 = dekripsi_affine(
    cipher_final, kunci_a, kunci_b
)

hasil2 = dekripsi_transposisi(
    hasil1, kunci_transposisi
)

hasil3 = dekripsi_vigenere(
    hasil2, kunci_vigenere
)

plaintext_kembali = dekripsi_caesar(
    hasil3, kunci_caesar
)

waktu_dekripsi = time.time() - waktu_mulai

print("Ciphertext         :", cipher_final)
print("Hasil Affine       :", hasil1)
print("Hasil Transposisi  :", hasil2)
print("Hasil Vigenere     :", hasil3)
print("Plaintext kembali  :", plaintext_kembali)
print("Waktu dekripsi     :", waktu_dekripsi, "detik")


# ==========================================
# 9. PEMERIKSAAN HASIL
# ==========================================

print("\n===================================")

if plaintext == plaintext_kembali:
    print("DEKRIPSI BERHASIL!")
    print("Plaintext berhasil dipulihkan.")
else:
    print("DEKRIPSI GAGAL!")

print("===================================")
```
